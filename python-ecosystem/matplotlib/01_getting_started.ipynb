# Matplotlib
Matplotlib is an open-source library for creating static, animated and interactive visualizations in Python.    
Its object-oriented API enables the embedding of plots into applications developed with GUI toolkits such as Tkinter, Qt and GTK.   
> It supports line plots, bar charts, histograms, scatter plots and 3D visualizations.  
> Provides full command over figure elements, including axes, labels, legends and colors.  
> Built to work with NumPy arrays and Pandas DataFrames.  
> Compatible with multiple general-purpose GUI toolkits (Tkinter, Qt, GTK and wxPython).   
> Capable of generating high-quality output in various hardcopy formats (PNG, PDF, SVG).

**Plotting**   
``plot()`` function is used to draw points (markers) in a diagram.   
The plot function will create a line plot connecting the points defined by these x and y coordinates.

```python
import matplotlib.pyplot as plt

x = [0, 2, 4, 6, 8]
y = [2, 5, 7, 8, 9]

plt.plot(x, y)
# Parameter 1 is an array of x values
# Parameter 2 is an array of y values
plt.show()
```

**Plotting without line**   
The marker parameter is used to specify the shape of the markers that will be drawn at each data point.     
In this case, ``o`` indicates that circular markers will be used.    

```python
x = [1, 2, 3, 4, 5]
y = [2, 5, 7, 8, 9]

plt.plot(x, y, "o") 
plt.show()
```

**Plotting with default values**   
If we do not specify the points on the x-axis, they will get the default values 0, 1, 2, 3 etc., depending on the length of the y-points.

```python
y = [2, 5, 1, 3, 10]

plt.plot(y)

plt.show()
```
