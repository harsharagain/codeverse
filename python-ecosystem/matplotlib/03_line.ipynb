# Line

**Linestyle**   
``linestyle`` (or) ``ls`` to set the style of the line connecting the markers    
> dotted can be written as :   
> dashed can be written as --   
> solid can be written as -    

**Line style reference:** https://matplotlib.org/stable/gallery/lines_bars_and_markers/linestyles.html#sphx-glr-gallery-lines-bars-and-markers-linestyles-py  


```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 5, 7, 8, 9]

plt.plot(x,y, ls="dotted")
plt.show()
```

**Linecolor**     
``color`` (or) ``c`` to set the color of the line connecting the markers

```python
y = [2, 5, 7, 8, 9]

plt.plot(x, y, color='g') # Set the color of the line to green
plt.show()
```

**Linewidth**   
``linewidth`` (or) ``lw`` to set the width of the line connecting the markers

```python
y = [2, 5, 7, 2, 9]

plt.plot(x, y, lw=20) # Set the width of the line to 20
plt.show()
```

**Multiple lines**    
Plot more lines using multiple plot() functions

```python
x = [1, 2, 3, 4, 5]
y1 = [2, 5, 7, 2, 9]
y2 = [1, 4, 6, 7, 2]

plt.plot(x, y1, color='b')
plt.plot(x, y2, color='y')

plt.show()
```
