**Sort**    
To sort an array, use ``sort()`` method   
This method returns a copy of the array by sorting it, leaving the original array unchanged.

```python
import numpy as np

arr = np.array([3, 2, 0, 1])
print(np.sort(arr))
```

Array can be sorted **alphabetically**

```python
arr = np.array(['papaya', 'pineapple', 'apple'])
print(np.sort(arr))
```

**Sorting boolean array**

```python
arr = np.array([True, False, True])
print(np.sort(arr))
```

**Sorting 2D Array**

```python
import numpy as np

arr = np.array([[3, 2, 4], [5, 0, 1]])
print(np.sort(arr))
```
