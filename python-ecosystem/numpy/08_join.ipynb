# Join
``concatenate()`` function allows us to join two or more arrays along an existing axis.

**1D Arrays**

```python
import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

arr = np.concatenate((arr1, arr2))
print(arr)
```

**2D Arrays**

```python
# Using ``axis=1``

arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])

arr = np.concatenate((arr1, arr2), axis=1)
print(arr) 
```

```python
# Without using (axis=1)

arr = np.concatenate((arr1, arr2))
print(arr)
```

**Stack**

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

arr = np.stack((arr1, arr2), axis=1) 
# Stacking along axis 1 means that the arrays will be stacked as columns.
print(arr)
```

```python
arr = np.stack((arr1, arr2))
print(arr)
```

**hstack()**   
hstack function is used to stack arrays in sequence horizontally (column wise).

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

arr = np.hstack((arr1, arr2))
print(arr)
```

**vstack()**    
vstack function is used to stack arrays in sequence vertically (row wise).

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

arr = np.vstack((arr1, arr2))
print(arr)
```

**dstack()**   
Stacking along height (depth)

``dstack()`` function is used to stack arrays in sequence depth wise (along third axis).
stacks along height, which is the same as depth.
