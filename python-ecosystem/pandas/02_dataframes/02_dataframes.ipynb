# DataFrames    
Data sets in Pandas are usually multi-dimensional tables, called DataFrames.    
Series is like a column whereas DataFrame is the whole table

```python
import pandas as pd

cart = {
    "cost":[450, 550, 300],
    "quantity": [2, 4, 5]
}

myvar = pd.DataFrame(cart)
print(myvar)
```

```python
cart = {
    "cost":[450, 550, 300],
    "quantity": [2, 4, 5]
}

myvar = pd.Series(cart)
print(myvar)
```

**Locate**    
Pandas use the ``loc`` attribute to return one or more specified row(s)

```python
cart = {
    "price": [560, 430, 120],
    "quantity": [2, 3, 1]
}
myvar = pd.DataFrame(cart)
print(myvar.loc[0]) # Return row (0th index)
print("-")
print(myvar.loc[[0, 1]]) # Return both row 0 and row 1
```

**Named Indexes**   
With the index argument, you can name your own indexes.

```python
data = {
    "price": [300, 400, 500 ],
    "quantity": [ 2, 1, 1 ]
}

myvar = pd.DataFrame(data, index=['bottle', 'bag1','bag2'])
print(myvar)
```

**Locate Named Indexes**   
Use the named index in the loc attribute to return the specified row(s).

```python
data = {
    "price": [300, 400, 500 ],
    "quantity": [ 2, 1, 1 ]
}

myvar = pd.DataFrame(data, index=['bottle', 'bag1','bag2'])
print(myvar.loc['bag2'])
print("-")
print(myvar.loc[["bottle","bag1"]])
```
